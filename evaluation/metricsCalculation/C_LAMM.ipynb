```python
import pandas as pd
import matplotlib.pyplot as plt
from metricsCalculation import DataPreparationUtils as utils
```

```python
listOf1ServerLogs = {
    '../data/1_server/5_clients': 5,
    '../data/1_server/10_clients': 10,
    '../data/1_server/15_clients': 15,
    '../data/1_server/20_clients': 20,
    '../data/1_server/25_clients': 25,
    '../data/1_server/30_clients': 30,
}

listOf3ServerLogs = {
    '../data/3_server/5_clients': 5,
    '../data/3_server/10_clients': 10,
    '../data/3_server/15_clients': 15,
    '../data/3_server/20_clients': 20,
    '../data/3_server/25_clients': 25,
    '../data/3_server/30_clients': 30,
}

listOf5ServerLogs = {
    '../data/5_server/5_clients': 5,
    '../data/5_server/10_clients': 10,
    '../data/5_server/15_clients': 15,
    '../data/5_server/20_clients': 20,
    '../data/5_server/25_clients': 25,
    '../data/5_server/30_clients': 30,
}
```

```python
dfCLAMM1Server = pd.DataFrame(columns=['Number of Mobile Sensing Clients', 'C_LAMM (Median)'])

for (logPath, clientNumber) in listOf1ServerLogs.items():
    dfLog = utils.readInClientLogsForScenario(logPath)[0]
    dfLogDescribe = dfLog.describe()
    median = dfLogDescribe.iat[5,2]

    dfCLAMM1Server.loc[len(dfCLAMM1Server)] = [clientNumber, median]

dfCLAMM3Server = pd.DataFrame(columns=['Number of Mobile Sensing Clients', 'C_LAMM (Median)'])

for (logPath, clientNumber) in listOf3ServerLogs.items():
    dfLog = utils.readInClientLogsForScenario(logPath)[0]
    dfLogDescribe = dfLog.describe()
    median = dfLogDescribe.iat[5,2]

    dfCLAMM3Server.loc[len(dfCLAMM3Server)] = [clientNumber, median]

dfCLAMM5Server = pd.DataFrame(columns=['Number of Mobile Sensing Clients', 'C_LAMM (Median)'])

for (logPath, clientNumber) in listOf5ServerLogs.items():
    dfLog = utils.readInClientLogsForScenario(logPath)[0]
    dfLogDescribe = dfLog.describe()
    median = dfLogDescribe.iat[5,2]

    dfCLAMM5Server.loc[len(dfCLAMM5Server)] = [clientNumber, median]
```

```python
figure = plt.figure()

plt.plot(dfCLAMM1Server['Number of Mobile Sensing Clients'], dfCLAMM1Server['C_LAMM (Median)'], label='1 Server')
plt.plot(dfCLAMM3Server['Number of Mobile Sensing Clients'], dfCLAMM3Server['C_LAMM (Median)'], label='3 Server')
plt.plot(dfCLAMM5Server['Number of Mobile Sensing Clients'], dfCLAMM5Server['C_LAMM (Median)'], label='5 Server')

plt.title("C_LAMM in Dependency of the Number of Server Instances and Clients")
plt.xlabel("Number of Mobile Sensing Clients")
plt.ylabel("C_LAMM in Milliseconds (Median)")

plt.grid()
plt.legend()

figure.set_figwidth(10)
figure.set_figheight(5)

plt.show()
```
