```python
import pandas as pd
import matplotlib.pyplot as plt
from metricsCalculation import DataPreparationUtils as utils
```

```python
listOf1ServerLogs = {
    '../data/1_server/5_clients': 5,
    '../data/1_server/10_clients': 10,
    '../data/1_server/15_clients': 15,
    '../data/1_server/20_clients': 20,
    '../data/1_server/25_clients': 25,
    '../data/1_server/30_clients': 30,
}

listOf3ServerLogs = {
    '../data/3_server/5_clients': 5,
    '../data/3_server/10_clients': 10,
    '../data/3_server/15_clients': 15,
    '../data/3_server/20_clients': 20,
    '../data/3_server/25_clients': 25,
    '../data/3_server/30_clients': 30,
}

listOf5ServerLogs = {
    '../data/5_server/5_clients': 5,
    '../data/5_server/10_clients': 10,
    '../data/5_server/15_clients': 15,
    '../data/5_server/20_clients': 20,
    '../data/5_server/25_clients': 25,
    '../data/5_server/30_clients': 30,
}
```

```python
dfSFAMM1Server = pd.DataFrame(columns=['Number of Mobile Sensing Clients', 'S_FAMM (Median)'])

for (logPath, clientNumber) in listOf1ServerLogs.items():
    dfLog = utils.readInServerLogsForScenario(logPath)[1]
    dfLogDescribe = dfLog.describe()
    median = dfLogDescribe.iat[5,0]

    dfSFAMM1Server.loc[len(dfSFAMM1Server)] = [clientNumber, median]

dfSFAMM3Server = pd.DataFrame(columns=['Number of Mobile Sensing Clients', 'S_FAMM (Median)'])

for (logPath, clientNumber) in listOf3ServerLogs.items():
    dfLog = utils.readInServerLogsForScenario(logPath)[1]
    dfLogDescribe = dfLog.describe()
    median = dfLogDescribe.iat[5,0]

    dfSFAMM3Server.loc[len(dfSFAMM3Server)] = [clientNumber, median]

dfSFAMM5Server = pd.DataFrame(columns=['Number of Mobile Sensing Clients', 'S_FAMM (Median)'])

for (logPath, clientNumber) in listOf5ServerLogs.items():
    dfLog = utils.readInServerLogsForScenario(logPath)[1]
    dfLogDescribe = dfLog.describe()
    median = dfLogDescribe.iat[5,0]

    dfSFAMM5Server.loc[len(dfSFAMM5Server)] = [clientNumber, median]
```

```python
figure = plt.figure()

plt.plot(dfSFAMM1Server['Number of Mobile Sensing Clients'], dfSFAMM1Server['S_FAMM (Median)'], label='1 Server')
plt.plot(dfSFAMM3Server['Number of Mobile Sensing Clients'], dfSFAMM3Server['S_FAMM (Median)'], label='3 Server')
plt.plot(dfSFAMM5Server['Number of Mobile Sensing Clients'], dfSFAMM5Server['S_FAMM (Median)'], label='5 Server')

plt.title("S_FAMM in Dependency of the Number of Server Instances and Clients")
plt.xlabel("Number of Mobile Sensing Clients")
plt.ylabel("S_FAMM in Milliseconds (Median)")

plt.grid()
plt.legend()

figure.set_figwidth(10)
figure.set_figheight(5)

plt.show()
```
